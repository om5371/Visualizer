```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python

class SalesDataAnalyzer:

    def __init__(self):
        self.df = None
        self.last_fig = None

    # ================= LOAD DATASET =================
    def load_dataset(self, file_path):
        try:
            self.df = pd.read_csv(file_path)
            print("Dataset loaded successfully!")

        except FileNotFoundError:
            print("Error: File not found. Please check the path and try again.")

        except Exception as e:
            print(f"Something went wrong while loading the dataset: {e}")

    # ================= CHECK DATASET =================
    def _check_loaded(self):
        if self.df is None:
            print("No dataset loaded yet! Please choose option 1 first.")
            return False

        return True

    # ================= EXPLORE DATA =================
    def explore_data(self):

        if not self._check_loaded():
            return

        print("\n== Explore Data ==")
        print("1. Display the first 5 rows")
        print("2. Display the last 5 rows")
        print("3. Display column names")
        print("4. Display data types")
        print("5. Display basic info")

        try:
            choice = int(input("\nEnter your choice: "))

            match choice:

                case 1:
                    print()
                    print(self.df.head())

                case 2:
                    print()
                    print(self.df.tail())

                case 3:
                    print()
                    print(list(self.df.columns))

                case 4:
                    print()
                    print(self.df.dtypes)

                case 5:
                    print()
                    self.df.info()

                case _:
                    print("Invalid choice.")

        except Exception as e:
            print(f"Error: {e}")

    # ================= DATAFRAME OPERATIONS =================
    def dataframe_operations(self):

        if not self._check_loaded():
            return

        print("\n== DataFrame Operations ==")
        print("1. Calculate Total Sales")
        print("2. Add New Column")
        print("3. Sort Data by Sales")
        print("4. Filter Sales Greater Than a Value")
        print("5. Search Product")

        try:
            choice = int(input("Enter your choice: "))

            match choice:

                case 1:
                    total = self.df["Sales"].sum()
                    print(f"\nTotal Sales: {total}")

                case 2:
                    col_name = input("Enter new column name: ")
                    default_value = input(
                        f"Enter default value for '{col_name}': "
                    )

                    self.df[col_name] = default_value

                    print(
                        f"\nColumn '{col_name}' added successfully!"
                    )

                case 3:
                    print(
                        self.df.sort_values(
                            by="Sales",
                            ascending=False
                        )
                    )

                case 4:
                    value1 = int(input("Enter the sales value: "))

                    print(
                        self.df[self.df["Sales"] > value1]
                    )

                case 5:
                    product = input(
                        "Enter product name to search: "
                    )

                    result = self.df[
                        self.df["Product"].str.contains(
                            product,
                            case=False,
                            na=False
                        )
                    ]

                    if not result.empty:
                        print(result)
                    else:
                        print("No matching product found.")

                case _:
                    print("Invalid choice.")

        except KeyError as e:
            print(f"Column not found: {e}")

        except ValueError:
            print("Please enter a valid value.")

        except Exception as e:
            print(f"Error: {e}")

    # ================= HANDLE MISSING DATA =================
    def handle_missing_data(self):

        if not self._check_loaded():
            return

        print("\n== Handle Missing Data ==")
        print("1. Display rows with missing values")
        print("2. Fill missing values with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing values with a specific value")

        try:
            choice = int(input("Enter your choice: "))

            match choice:

                case 1:
                    missing_rows = self.df[
                        self.df.isnull().any(axis=1)
                    ]

                    if missing_rows.empty:
                        print(
                            "\nNo missing values found in the dataset!"
                        )
                    else:
                        print(missing_rows)

                case 2:
                    numeric_cols = self.df.select_dtypes(
                        include="number"
                    ).columns

                    self.df[numeric_cols] = self.df[
                        numeric_cols
                    ].fillna(
                        self.df[numeric_cols].mean()
                    )

                    print(
                        "\nMissing numeric values filled "
                        "with column mean!"
                    )

                case 3:
                    before = len(self.df)

                    self.df = self.df.dropna()

                    print(
                        f"\nDropped {before - len(self.df)} "
                        "rows with missing values!"
                    )

                case 4:
                    value = input(
                        "Enter the value to replace "
                        "missing data with: "
                    )

                    self.df = self.df.fillna(value)

                    print(
                        "\nMissing values replaced successfully!"
                    )

                case _:
                    print("Invalid choice.")

        except ValueError:
            print("Please enter a valid number.")

        except Exception as e:
            print(f"Error: {e}")

    # ================= DESCRIPTIVE STATISTICS =================
    def descriptive_statistics(self):

        if not self._check_loaded():
            return

        print("\n== Descriptive Statistics ==")
        print("1. Display Descriptive Statistics")
        print("2. Calculate Average Sales")
        print("3. Calculate Maximum Sales")
        print("4. Calculate Minimum Sales")
        print("5. Region-wise Total Sales")

        try:
            choice = int(input("Enter your choice: "))

            match choice:

                case 1:
                    print(self.df.describe())

                case 2:
                    print(
                        f"Average Sales: "
                        f"{self.df['Sales'].mean():.2f}"
                    )

                case 3:
                    print(
                        f"Maximum Sales: "
                        f"{self.df['Sales'].max()}"
                    )

                case 4:
                    print(
                        f"Minimum Sales: "
                        f"{self.df['Sales'].min()}"
                    )

                case 5:
                    print(
                        self.df.groupby("Region")["Sales"].sum()
                    )

                case _:
                    print("Invalid choice.")

        except KeyError as e:
            print(f"Column not found: {e}")

        except ValueError:
            print("Please enter a valid number.")

        except Exception as e:
            print(f"Error: {e}")

    # ================= DATA VISUALIZATION =================
    def data_visualization(self):

        if not self._check_loaded():
            return

        print("\n== Data Visualization ==")
        print("1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Chart")
        print("5. Histogram")
        print("6. Stack Plot")

        try:
            choice = int(input("Enter your choice: "))

            match choice:

                # ---------- BAR PLOT ----------
                case 1:
                    x_col = input(
                        "Enter x-axis column name: "
                    )

                    y_col = input(
                        "Enter y-axis column name: "
                    )

                    print("Generating bar plot...")

                    self.last_fig = plt.figure()

                    sns.barplot(
                        data=self.df,
                        x=x_col,
                        y=y_col
                    )

                    plt.title(
                        f"{y_col} by {x_col}"
                    )

                    plt.show()

                    print(
                        "Bar plot displayed successfully!"
                    )

                # ---------- LINE PLOT ----------
                case 2:
                    x_col = input(
                        "Enter x-axis column name: "
                    )

                    y_col = input(
                        "Enter y-axis column name: "
                    )

                    print("Generating line plot...")

                    self.last_fig = plt.figure()

                    plt.plot(
                        self.df[x_col],
                        self.df[y_col],
                        marker="o"
                    )

                    plt.xlabel(x_col)
                    plt.ylabel(y_col)

                    plt.title(
                        f"{y_col} over {x_col}"
                    )

                    plt.show()

                    print(
                        "Line plot displayed successfully!"
                    )

                # ---------- SCATTER PLOT ----------
                case 3:
                    x_col = input(
                        "Enter x-axis column name: "
                    )

                    y_col = input(
                        "Enter y-axis column name: "
                    )

                    print("Generating scatter plot...")

                    self.last_fig = plt.figure()

                    plt.scatter(
                        self.df[x_col],
                        self.df[y_col]
                    )

                    plt.xlabel(x_col)
                    plt.ylabel(y_col)

                    plt.title(
                        f"{y_col} vs {x_col}"
                    )

                    plt.show()

                    print(
                        "Scatter plot displayed successfully!"
                    )

                # ---------- PIE CHART ----------
                case 4:
                    cat_col = input(
                        "Enter category column name: "
                    )

                    val_col = input(
                        "Enter values column name: "
                    )

                    print("Generating pie chart...")

                    grouped = self.df.groupby(
                        cat_col
                    )[val_col].sum()

                    self.last_fig = plt.figure()

                    plt.pie(
                        grouped,
                        labels=grouped.index,
                        autopct="%1.1f%%"
                    )

                    plt.title(
                        f"{val_col} share by {cat_col}"
                    )

                    plt.show()

                    print(
                        "Pie chart displayed successfully!"
                    )

                # ---------- HISTOGRAM ----------
                case 5:
                    col = input(
                        "Enter column name for histogram: "
                    )

                    print("Generating histogram...")

                    self.last_fig = plt.figure()

                    plt.hist(
                        self.df[col],
                        bins=10,
                        edgecolor="black"
                    )

                    plt.xlabel(col)
                    plt.ylabel("Frequency")

                    plt.title(
                        f"Distribution of {col}"
                    )

                    plt.show()

                    print(
                        "Histogram displayed successfully!"
                    )

                # ---------- STACK PLOT ----------
                case 6:
                    cat_col = input(
                        "Enter category column name: "
                    )

                    val_col = input(
                        "Enter values column name: "
                    )

                    print("Generating stack plot...")

                    grouped = self.df.groupby(
                        cat_col
                    )[val_col].sum()

                    self.last_fig = plt.figure()

                    plt.stackplot(
                        grouped.index,
                        grouped.values
                    )

                    plt.xlabel(cat_col)
                    plt.ylabel(val_col)

                    plt.title(
                        f"{val_col} Stack Plot by {cat_col}"
                    )

                    plt.show()

                    print(
                        "Stack plot displayed successfully!"
                    )

                case _:
                    print("Invalid choice.")

        except KeyError as e:
            print(f"Column not found: {e}")

        except ValueError:
            print("Please enter a valid number.")

        except Exception as e:
            print(
                f"Error creating plot: {e}"
            )

    # ================= SAVE VISUALIZATION =================
    def save_visualization(self):

        try:
            if self.last_fig is None:
                print(
                    "No visualization to save yet! "
                    "Create one first (option 6)."
                )
                return

            file_name = input(
                "Enter file name to save the plot "
                "(e.g., scatter_plot.png): "
            )

            os.makedirs(
                "visualizations",
                exist_ok=True
            )

            save_path = os.path.join(
                "visualizations",
                file_name
            )

            self.last_fig.savefig(save_path)

            print(
                f"Visualization saved as "
                f"{save_path} successfully!"
            )

        except Exception as e:
            print(
                f"Error saving visualization: {e}"
            )


# ================= MAIN FUNCTION =================
def main():

    analyzer = SalesDataAnalyzer()

    while True:

        print(
            "\n========== Data Analysis & Visualization Program =========="
        )

        print("Please select an option:")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Generate Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Visualization")
        print("8. Exit")

        print(
            "================================================================="
        )

        try:
            choice = int(
                input("\nEnter your choice: ")
            )

        except ValueError:
            print(
                "Please enter a valid number (1-8)."
            )
            continue

        try:

            match choice:

                case 1:
                    print("\n== Load Dataset ==")

                    path = input(
                        "Enter the path of the dataset "
                        "(CSV file): "
                    )

                    analyzer.load_dataset(path)

                case 2:
                    analyzer.explore_data()

                case 3:
                    analyzer.dataframe_operations()

                case 4:
                    analyzer.handle_missing_data()

                case 5:
                    analyzer.descriptive_statistics()

                case 6:
                    analyzer.data_visualization()

                case 7:
                    analyzer.save_visualization()

                case 8:
                    print(
                        "\nExiting the program. Goodbye!"
                    )
                    break

                case _:
                    print(
                        "Invalid choice, please select "
                        "between 1 and 8."
                    )

        except Exception as e:
            print(
                f"Unexpected error: {e}"
            )


main()
```
